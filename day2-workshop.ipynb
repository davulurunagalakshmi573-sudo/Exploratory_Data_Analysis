```python
#Step 1 importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
sns.set_style(style='whitegrid') #to sey bg style of graphs
```

```python
#Step 2 Loading (autompg dataset) and cleaning the Data
data = fetch_ucirepo(id=9)
data
x=data.data.features.copy() # returns all numerical features
x.describe()
x.info()
x
y=data.data.targets['mpg']
y
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python
#first we will transform the horsepower column and horese power and origin columns
x['horsepower']=pd.to_numeric(x['horsepower'],errors='coerce')
x['horsepower'].info()
x['origin']=x['origin'].astype(str)
```

```python
#all numerical columns into a single collection
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features=['origin']
x=x[numerical_features+categorical_features]
x.info()
x.isnull().sum()
x.shape
```

```python
#Step 3 some quick exploratory data analysis where we use seaborn
df=x.copy()
df['mpg']=y
#here we are mapping input and target variable into the dataframe
fig,axes=plt.subplots(1,3,figsize=(12,4))
#plot graphs as per axes
sns.histplot(df['mpg'], kde=True, ax=axes[0],color='red')
#in similar way we display other graphs in rmng plots
axes[0].set_title("MPG Distribution")
sns.scatterplot(x=df['weight'],y=df['mpg'],ax=axes[1])
axes[1].set_title("Weight vs MPG Distribution")
sns.scatterplot(x=df['horsepower'],y=df['mpg'],ax=axes[2],color='yellow')
axes[2].set_title("HorsePower vs MPG Distribution")
```

```python
#correlation between our variables(positive correlation/negative correlation)
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),annot=True,cmap='coolwarm',center=0)
plt.title("Correlation Matric")
plt.tight_layout()
plt.show()
```

```python
#Step 4 Dividing the data into train,test and split
#print(x)
#print(y)
#split rule -->70-30,80-20,75-25
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('\n Train test split')
print("Training Data",len(x_train))
print("Testing Data",len(x_test))
#print(x_train)
```

```python
#Step 5 :Creating a Evaluate function baseline Model
def evaluate(model_name,actual,predicted):
    """Returns MAE,MSE,RMSE,R2-Score"""
    mae=mean_absolute_error(actual,predicted)
    mse=mean_squared_error(actual,predicted)
    return{'Model_Name':model_name,
    'MAE':mae,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-Score':r2_score(actual,predicted)}
results=[]
```

```python
#Step 6 : Baseline model -->It predicts the training set for every value
baseline_prediction=np.full(len(y_test),y_train.mean())
#baseline_predictions
#in above case we are keeping a baseline prediction
results.append(evaluate("Mean baseline",y_test,baseline_prediction))
#results
#in above case we are creating the baseline predicted data
```

```python
#step 7 :we apply linear regression
simple_model=LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions=simple_model.predict(x_test[['weight']])
#print(simple_models_predicted)
results.append(evaluate("Simple Linear Regression:weight",
                        y_test,
                        simple_model_predictions))
print(len(results))
```

```python
#step 8:Multiple Linear Regression
#Multiple Regeression uses one or more independent variables to fit with a linear plane
#Polynomial Regression uses powers of a single variable to fit a curved line,though it is 
#Calculated using linear equation...

#we have missing numeric values in horsepower column we will fill it
#we have origin column which is category -->future encoding
preprocessor=ColumnTransformer(transformers=[
    ("numbers",SimpleImputer(strategy="median"),
     numerical_features),
    ("category",Pipeline(steps=[("imputer",SimpleImputer(strategy="most_frequent")),
                                 ("encoder",OneHotEncoder(handle_unknown="ignore",
                                                          drop="first"),
                                 )]),
    categorical_features),
]
)
preprocessor
numerical_features
```

```python
multiple_model=Pipeline(
    steps=[("prepare",preprocessor),
           ("model",LinearRegression())]
)
multiple_model.fit(x_train,y_train)
```

```python
multiple_predictions=multiple_model.predict(x_test)
results.append(evaluate("Multiple Regression",y_test,multiple_predictions))
results
```

```python
# Step 9: Polynomial Regression
# It forms a curvilinear relationship
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV
import pandas as pd

# Define the pipeline steps correctly targeting sub-parameters
polynomial_model = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("polynomial", PolynomialFeatures(include_bias=False)),
    ("model", LinearRegression())
])

# Selecting the polynomial degree (using double underscores '__' to reach inside the pipeline)
degree_search = GridSearchCV(
    polynomial_model,
    param_grid={'polynomial__degree': [1, 2, 3, 4, 5]},
    scoring="neg_mean_squared_error",
    cv=5
)

# Fit the grid search model
degree_search.fit(x_train[['horsepower']], y_train)

# Generate predictions
polynomial_predictions = degree_search.predict(x_test[['horsepower']])

# Modern Pandas replacement for .append(): Create a single-row DataFrame
new_row = pd.DataFrame([{
    "Model": "Polynomial :Horse Power",
    "Actual": y_test.values,
    "Predicted": polynomial_predictions
}])

# Combine the new row into your existing 'results' DataFrame
results = pd.concat([results, new_row], ignore_index=True)

# Display the updated results table
results
```

```python
#Findout best degree
best_degree=degree_search.best_params_['polynomial__degree']
cv_rmse=np.sqrt(-degree_search.best_score_)
print(best_degree)
print(cv_rmse)
```

```python
#finally we will print the results
results=pd.DataFrame(results).sort_values("RMSE")
results
```
