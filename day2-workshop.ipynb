```python
#importing the libraries
import pandas as pd
import numpy as np
import sklearn
```

```python
data = pd.DataFrame({'Price_USA':[80,50,140,25],
                    'Price_INR':[5000,7000,12000,3000]})
data
```

```python
#We will perform mathematical wise
data['Price_USA_minmax']=(data['Price_USA']-data['Price_USA'].min())/(data['Price_USA'].max()-data['Price_USA'].min())
data
```

```python
#We will perform mathematical wise
data['Price_INR_minmax']=(data['Price_INR']-data['Price_INR'].min())/(data['Price_INR'].max()-data['Price_INR'].min())
data
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
min_max=MinMaxScaler()
```

```python
#we need to use fit_transform()
data['Price_USA_minsk']=min_max.fit_transform(data[['Price_USA']])
data
```

```python
#we need to use fit_transform()
data['Price_INR_minsk']=min_max.fit_transform(data[['Price_INR']])
data
```

```python
#Standardadization (StandardScaler)=x(i)-mean/std
data['Price_USA_std']=(data['Price_USA']-data['Price_USA'].mean())/(data['Price_USA'].std())
data
```

```python
#Standardadization (StandardScaler)=x(i)-mean/std
data['Price_INR_std']=(data['Price_INR']-data['Price_INR'].mean())/(data['Price_INR'].std())
data
```

```python
#Step 1 importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
sns.set_style(style='whitegrid') #to sey bg style of graphs
```

```python
#Step 2 Loading (autompg dataset) and cleaning the Data
data = fetch_ucirepo(id=9)
data
x=data.data.features.copy() # returns all numerical features
x.describe()
x.info()
x
y=data.data.targets['mpg']
y
```

```python
#first we will transform the horsepower column and horese power and origin columns
x['horsepower']=pd.to_numeric(x['horsepower'],errors='coerce')
x['horsepower'].info()
x['origin']=x['origin'].astype(str)
```

```python
#all numerical columns into a single collection
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features=['origin']
x=x[numerical_features+categorical_features]
x.info()
x.isnull().sum()
x.shape
```

```python
#Step 3 some quick exploratory data analysis where we use seaborn
df=x.copy()
df['mpg']=y
#here we are mapping input and target variable into the dataframe
fig,axes=plt.subplots(1,3,figsize=(12,4))
#plot graphs as per axes
sns.histplot(df['mpg'],kde=False,ax=axes[0],color='red')
#in similar way we display other graphs in rmng plots
axes[0].set_title("MPG Distribution")
sns.scatterplot(x=df['weight'],y=df['mpg'],ax=axes[1])
axes[1].set_title("Weight vs MPG Distribution")
sns.scatterplot(x=df['horsepower'],y=df['mpg'],ax=axes[2],color='yellow')
axes[2].set_title("HorsePower vs MPG Distribution")
```

```python
#Step 4 Dividing the data into train,test and split
#print(x)
#print(y)
#split rule -->70-30,80-20,75-25
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print('\n Train test split')
print("Training Data",len(x_train))
print("Testing Data",len(x_test))
#print(x_train)
```

```python
#Step 5 :Creating a Evaluate function baseline Model
def evaluate(model_name,actual,predicted):
    """Returns MAE,MSE,RMSE,R2-Score"""
    mae=mean_absolute_error(actual,predicted)
    mse=mean_squared_error(actual,predicted)
    return{'Model_Name':model_name,
    'MAE':mae,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-Score':r2_score(actual,predicted)}
results=[]
```

```python
#Step 6 : Baseline model -->It predicts the training set for every value
baseline_prediction=np.full(len(y_test),y_train.mean())
#baseline_predictions
#in above case we are keeping a baseline prediction
results.append(evaluate("Mean baseline",y_test,baseline_prediction))
#results
#in above case we are creating the baseline predicted data
```

```python
#step 7 :we apply linear regression
simple_model=LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions=simple_model.predict(x_test[['weight']])
#print(simple_models_predicted)
results.append(evaluate("Simple Linear Regression",
                        y_test,
                        simple_model_predictions))
print(len(results))
```
